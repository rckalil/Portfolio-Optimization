```python
import yfinance as yf
import pandas as pd
import numpy as np

tickers = ["USDBRL=X", "EURBRL=X", "GBPBRL=X"]
# help(yf.download)
prices = yf.download(tickers, start="2022-01-01", end="2026-01-01")["Close"]
print(prices.head())
```

```python
returns = prices.pct_change().dropna()
print(returns.head())

log_returns = np.log(prices / prices.shift(1)).dropna()

mu_daily = returns.mean()
Sigma_daily = returns.cov()

mu_annual = mu_daily * 252
Sigma_annual = Sigma_daily * 252

mu = mu_annual.values
Sigma = Sigma_annual.values
```
